# Day 22: MLOps Basics with MLflow & Experiment Tracking & Versioning

**Week 4 ||  Advanced & Real-World ML**

Welcome to Day 22! Today you’ll learn the practical foundation of MLOps: **experiment tracking** and **basic model versioning** with **MLflow**. We’ll keep everything minimal and production‑minded so you can adopt these practices immediately.

---

## Plan
- Understand MLflow core concepts: Experiments, Runs, Parameters, Metrics, Artifacts, Models
- Track multiple model runs (hyperparameters, metrics, plots) using autologging + custom logs
- Compare runs in the MLflow UI and pick the best
- Save and load a tracked model for quick inference

## Learning Objectives

By the end of today, you will be able to:

1. Set up and use MLflow locally for experiment tracking
2. Log parameters, metrics, and artifacts (e.g., confusion matrix)
3. Run controlled experiments and compare results in the UI
4. Save and load models from MLflow for inference

### What You’ll Practice

- Minimal, repeatable training runs (each training = one run)
- Autologging vs. manual logging
- Clean artifacts (plots) and clear metrics
- Selecting the best run programmatically
